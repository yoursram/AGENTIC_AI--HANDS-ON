```python
# Why Ollama Works With a REST API
# Ollama relies on a central REST API server
# architecture for several critical engineering and architectural
#  reasons:
# 1.Large Language Models (LLMs) require massive computing resources,
# dedicated RAM, and heavy GPU acceleration. Decoupling the heavy engine (the server) 
# from the lightweight user interface (the CLI tool or web browser)
# ensures your computer's terminal doesn't freeze or drop the model from
# memory every time you press Enter
# 2.Stateful Model Management: Standard LLMs take a long time to boot up and
# load into VRAM. Ollama uses the background API server to keep models loaded 
# or efficiently swap them out based on incoming HTTP requests, acting as an 
# automated traffic controller for your local system memory
```

```python
# How It Works (The Lifecycle of a Request):


# [ User UI / CLI Tool / Code ] --> 
# [ Ollama REST API Server ]  <-- Runs in background on localhost:11434 ((1) Sends standard HTTP POST with JSON data)--->  
# [ Model Manager Engine ]    <-- Manages VRAM & switches active models ((2) Parses JSON & routes to Model Manager Queue) --->
# [ GPU / CPU Inference ]   <-- Processes mathematical weights  ((3) Feeds raw numerical tensors into hardware)
```

```python
#using wrappers
import ollama

response = ollama.generate(
    model="gemma4:12b",
    prompt="Explain black hole simply" 
)
print(response['response'])
```

```python
#generate using REST API
import requests
import json

url = "http://localhost:11434/api/generate"

payload ={
    "model":"gemma4:12b",
    "prompt":"explain  black holes in simply"
}
response =  requests.post(url,json=payload)
print(response)
```

```python
output = ""
for i in response.iter_lines():
    if i:
        data = json.loads(i.decode("utf-8"))
        if "response" in data:
            output += data["response"]
        if data.get("done"):
            break
print(output)
```
